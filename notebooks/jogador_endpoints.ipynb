# Funcionamento da entidade Jogador

Os principais endpoints são:
- Criar jogador
- Listar jogadores
- Obter jogador por ID
- Atualizar jogador
- Deletar jogador


## Criar jogador
**Endpoint:** `POST /jogadores/`

Este endpoint permite criar um novo jogador, associando-o a um clube existente. Se o clube informado não existir, será retornado um erro 404.

**Exemplo de requisição:**

```python
import requests

url = 'http://localhost:8000/jogadores/'
payload = {
    "nome": "Cauã de Freitas",
    "idade": 21,
    "posicao": "Atacante",
    "overall": 88,
    "clube_nome": "Flamengo"
}
response = requests.post(url, json=payload)
print(response.status_code)
print(response.json())
```

## Listar jogadores
**Endpoint:** `GET /jogadores/`

Retorna uma lista com todos os jogadores cadastrados na base de dados.

**Exemplo:**

```python
response = requests.get('http://localhost:8000/jogadores/')
print(response.status_code)
print(response.json())
```

## Obter jogador por ID
**Endpoint:** `GET /jogadores/{jogador_id}`

Retorna os dados de um jogador específico, identificado pelo seu ID.

**Exemplo:**

```python
jogador_id = 1
response = requests.get(f'http://localhost:8000/jogadores/{jogador_id}')
print(response.status_code)
print(response.json())
```

## Atualizar jogador
**Endpoint:** `PUT /jogadores/{jogador_id}`

Atualiza os dados de um jogador específico, incluindo o clube associado.

**Exemplo de requisição:**

```python
jogador_id = 1
payload = {
    "nome": "Cauã Freitas",
    "idade": 21,
    "posicao": "Meia",
    "overall": 99,
    "clube_nome": "Barcelona"
}
response = requests.put(f'http://localhost:8000/jogadores/{jogador_id}', json=payload)
print(response.status_code)
print(response.json())
```

## Deletar jogador
**Endpoint:** `DELETE /jogadores/{jogador_id}`

Remove um jogador da base de dados permanentemente.

**Exemplo:**

```python
jogador_id = 1
response = requests.delete(f'http://localhost:8000/jogadores/{jogador_id}')
print(response.status_code)
print(response.json())
```
