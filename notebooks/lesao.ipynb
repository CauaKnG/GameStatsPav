# Endpoints de Lesão

## Listar Lesões
Este endpoint retorna uma lista com todas as lesões cadastradas no banco de dados, incluindo o nome do jogador relacionado.

```python

import requests

response = requests.get("http://localhost:8000/lesoes/")
print(response.status_code)
print(response.json())
```

## Buscar Lesão por ID
Busca uma lesão específica com base no seu `id`. Se a lesão não existir, retorna erro 404.

```python

lesao_id = 1  
response = requests.get(f"http://localhost:8000/lesoes/{lesao_id}")
print(response.status_code)
print(response.json())
```

## Criar Nova Lesão
Cria um novo registro de lesão associado a um jogador (informado pelo nome).
- A duração estimada da lesão não pode ser negativa.
- A duração da lesão também não pode ser uma data futura.

```python

nova_lesao = {
    "nome_jogador": "Neymar Jr",
    "tipo_lesao": "Lesão muscular",
    "data_lesao": "2025-07-15",
    "duracao_estimada_dias": 30
}

response = requests.post("http://localhost:8000/lesoes/", json=nova_lesao)
print(response.status_code)
print(response.json())
```

## 4. Atualizar Lesão
Atualiza os dados de uma lesão existente com base no seu `id`. Pode-se alterar tipo de lesão, data, duração e jogador.

```python

lesao_id = 1  
dados_atualizados = {
    "nome_jogador": "Neymar Jr",
    "tipo_lesao": "Entorse no tornozelo",
    "data_lesao": "2025-07-15",
    "duracao_estimada_dias": 20
}

response = requests.put(f"http://localhost:8000/lesoes/{lesao_id}", json=dados_atualizados)
print(response.status_code)
print(response.json())
```

## 5. Deletar Lesão
Remove uma lesão do sistema com base no seu `id`. Retorna os dados da lesão removida.

```python

lesao_id = 1  
response = requests.delete(f"http://localhost:8000/lesoes/{lesao_id}")
print(response.status_code)
print(response.json())
```
