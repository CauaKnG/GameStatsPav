# Liga 
Este notebook demonstra e explica o funcionamento dos endpoints da entidade **Liga** na API desenvolvida com FastAPI.

## GET /ligas/
Lista todas as ligas cadastradas na base de dados.

```python
import requests
response = requests.get('http://localhost:8000/ligas/')
response.json()
```

## GET /ligas/{liga_id}
Busca uma liga específica pelo seu ID.
**404** se não encontrada.

```python
liga_id = 1
response = requests.get(f'http://localhost:8000/ligas/{liga_id}')
response.json()
```

## POST /ligas/
Cria uma nova liga.
Retorna os dados da liga criada com status **201**.

```python
nova_liga = {
    "nome": "Copa Libertadores",
    "pais": "América do Sul"
}
response = requests.post('http://localhost:8000/ligas/', json=nova_liga)
response.json()
```

## PUT /ligas/{liga_id}
Atualiza os dados de uma liga existente.
Se a liga não existir, retorna **404**.

```python
liga_id = 1
dados_atualizados = {
    "nome": "Brasileirão Série B",
    "pais": "Brasil"
}
response = requests.put(f'http://localhost:8000/ligas/{liga_id}', json=dados_atualizados)
response.json()
```

## DELETE /ligas/{liga_id}
Deleta uma liga pelo ID.
Se não encontrada, retorna **404**.
Se houver qualquer clube registrado na liga, retorna **400**.
Para deletar uma liga não poderá conter nenhum clube registrado nela.

```python
liga_id = 1
response = requests.delete(f'http://localhost:8000/ligas/{liga_id}')
response.json()
```
