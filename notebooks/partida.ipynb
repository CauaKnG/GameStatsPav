# Documentação dos Endpoints de Partida
Explicação detalhada dos endpoints relacionados à entidade **Partida**.

## Listar Partidas
**Rota:** `GET /`

Retorna a lista de todas as partidas cadastradas no banco de dados.

```python
import requests

response = requests.get("http://localhost:8000/partidas/")
response.json()
```

## Buscar Partida por ID
**Rota:** `GET /{partida_id}`

Busca uma partida específica pelo seu ID.

```python
partida_id = 1
response = requests.get(f"http://localhost:8000/partidas/{partida_id}")
response.json()
```

## Criar Partida
**Rota:** `POST /`

Cria uma nova partida. Os campos obrigatórios são: `data_partida`, `local`, `clube_casa`, `clube_fora`. Os campos `gols_casa` e `gols_fora` são opcionais.
- Não é permitido que os clubes de casa e fora de casa sejam iguais.
- Não é permitido criar partida com clubes inexistentes.

```python
partida_data = {
    "data_partida": "10/08/2025 15:30",
    "local": "Estádio do Maracanã",
    "clube_casa": "Flamengo",
    "clube_fora": "Vasco",
    "gols_casa": 1,
    "gols_fora": 2
}

response = requests.post("http://localhost:8000/partidas/", json=partida_data)
response.json()
```

## Atualizar Partida
**Rota:** `PUT /{partida_id}`

Atualiza os dados de uma partida existente.

```python
partida_id = 1
partida_atualizada = {
    "data_partida": "15/08/2025 18:00",
    "local": "Arena Corinthians",
    "clube_casa": "Corinthians",
    "clube_fora": "Palmeiras",
    "gols_casa": 0,
    "gols_fora": 0
}

response = requests.put(f"http://localhost:8000/partidas/{partida_id}", json=partida_atualizada)
response.json()
```

## Deletar Partida
**Rota:** `DELETE /{partida_id}`

Remove uma partida existente com base no seu ID.

```python
partida_id = 1
response = requests.delete(f"http://localhost:8000/partidas/{partida_id}")
response.json()
```
