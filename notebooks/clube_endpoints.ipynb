# Funcionamento da entidade Clube

Os principais endpoints são:
- Criar Clube
- Listar Clubes
- Obter Clube por ID
- Atualizar Clube
- Deletar Clube

## Requisitos

```python

import requests

BASE_URL = "http://localhost:8000/clubes"
    
```

## Listar todos os clubes

```python

response = requests.get(BASE_URL)
print("Status Code:", response.status_code)
response.json()
    
```

## Buscar um Clube em específico pelo seu ID

```python

clube_id = 1  # Altere conforme necessário
response = requests.get(f"{BASE_URL}/{clube_id}")
print("Status Code:", response.status_code)
response.json()
    
```

## Criar um novo clube

**Endpoint:** `POST /clubes/`

Este endpoint permite criar um novo clube, associando-o a uma liga existente. Se a liga informada não existir, será retornado um erro 404.

**Exemplo de requisição:**

```python

novo_clube = {
    "nome": "Flamengo",
    "cidade": "Rio de Janeiro",
    "nome_liga": "Brasileirão"
}
response = requests.post(BASE_URL, json=novo_clube)
print("Status Code:", response.status_code)
response.json()
    
```

## Atualizar um clube 

Atualiza um clube a partir do ID
Caso o nome da liga não seja correspondente a uma ja existente, retorna **400**.

```python

clube_id = 1  
atualizacao = {
    "nome": "Barcelona",
    "cidade": "Barcelona",
    "nome_liga": "La Liga"
}
response = requests.put(f"{BASE_URL}/{clube_id}", json=atualizacao)
print("Status Code:", response.status_code)
response.json()
    
```

## Deletar um clube existente a partir do ID

```python

clube_id = 1 
response = requests.delete(f"{BASE_URL}/{clube_id}")
print("Status Code:", response.status_code)
response.json()
    
```


## Observações
- O campo `nome_liga` ao criar um clube deve corresponder a uma liga já existente no banco de dados.
- A exclusão de clubes pode falhar se houver restrições de integridade (como jogadores vinculados a ele).
