
# Endpoints da Entidade Falta

Demonstração do funcionamento dos endpoints relacionados à entidade **Falta** do sistema.



## Listar Faltas

**Endpoint:** `GET /faltas/`  
**Descrição:** Retorna a lista de todas as faltas registradas no banco de dados.


```python

import requests

response = requests.get("http://localhost:8000/faltas/")
response.json()
```


## Buscar Falta por ID

**Endpoint:** `GET /faltas/{falta_id}`  
**Descrição:** Retorna os dados de uma falta específica, com base no seu ID.


```python

falta_id = 1
response = requests.get(f"http://localhost:8000/faltas/{falta_id}")
response.json()
```


## Criar Falta

**Endpoint:** `POST /faltas/`  

**Descrição:** Registra uma nova falta associada a um jogador e uma partida futura.

**Observações:** Minuto ocorrido da falta deve estar entre 0 e 120.


```python

nova_falta = {
    "nome_jogador": "Cauã Freitas",
    "minuto_ocorrido": 23,
    "tipo_cartao": "amarelo",
    "descricao": "Entrada dura no adversário",
    "dentro_area": False
}

response = requests.post("http://localhost:8000/faltas/", json=nova_falta)
response.json()
```


## Atualizar Falta

**Endpoint:** `PUT /faltas/{falta_id}`  
**Descrição:** Atualiza as informações de uma falta existente.


```python

falta_id = 1
atualizacao = {
    "nome_jogador": "Cauã Freitas",
    "minuto_ocorrido": 30,
    "tipo_cartao": "vermelho",
    "descricao": "Reclamação acintosa com o árbitro",
    "dentro_area": True
}

response = requests.put(f"http://localhost:8000/faltas/{falta_id}", json=atualizacao)
response.json()
```


## Deletar Falta

**Endpoint:** `DELETE /faltas/{falta_id}`  
**Descrição:** Remove uma falta do banco de dados.


```python

falta_id = 1
response = requests.delete(f"http://localhost:8000/faltas/{falta_id}")
response.json()
```
