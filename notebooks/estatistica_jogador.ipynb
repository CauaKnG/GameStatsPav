## Estatísticas dos Jogadores por partida
Este notebook demonstra como interagir com os endpoints da API relacionados à entidade **EstatísticasJogador**.

```python
import requests
import json
```

```python
BASE_URL = "http://localhost:8000/estatisticas_jogador"
BASE_URL_JOGADOR = "http://localhost:8000/jogadores"
BASE_URL_PARTIDA = "http://localhost:8000/partidas"
```

## Listar todas as estatísticas de jogadores

```python
response = requests.get(BASE_URL)
print(response.status_code)
print(response.json())
```

## Buscar estatísticas por jogador (por ID do jogador)

```python
jogador_id = 1
response = requests.get(f"{BASE_URL}/{jogador_id}")
print(response.status_code)
print(response.json())
```

## Criar estatísticas para um jogador

- Caso o número de gols feitos pelos jogadores exceda o número total de gols da equipe na partida, retorna **400**.
- Caso o clube daquele jogador não tenha participado da partida, retorna **400**.
- Caso o jogador não esteja em nenhum dos dois clubes que participaram da partida, retorna **400**.
- O valor das estatísticas não podem ser negativos.

```python
nova_estatistica = {
    "nome_jogador": "Cauã Freitas",
    "gols": 0,
    "assistencias": 0,
    "passes_completos": 35,
    "finalizacoes": 5,
    "partida_id": 1
}

response = requests.post(BASE_URL, json=nova_estatistica)
print(response.status_code)
print(response.json())
```

## Atualizar estatísticas de um jogador

- Caso número de gols feitos pelos jogadores exceda o número total de gols da equipe na partida, retorna **400**.
- Caso o clube daquele jogador não tenha participado da partida, retorna **400**.
- Caso o jogador não esteja em nenhum dos dois clubes que participaram da partida, retorna **400**.

```python
estatistica_id = 1 
atualizacao = {
    "nome_jogador": "Cauã Freitas",
    "gols": 1,
    "assistencias": 2,
    "passes_completos": 40,
    "finalizacoes": 4,
    "partida_id": 1
}
response = requests.put(f"{BASE_URL}/{estatistica_id}", json=atualizacao)
print(response.status_code)
print(response.json())
```

## Deletar estatísticas de um jogador

```python
estatisticas_id = 1  
response = requests.delete(f"{BASE_URL}/{estatisticas_id}")
print(response.status_code)
print(response.json())
```
